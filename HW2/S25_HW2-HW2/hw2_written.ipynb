```python
import numpy as np
```

```python
X_train = np.array([[1, 1],
                    [1, 0],
                    [0, 1],
                    [2, 1]])

Y_train = np.array([1, 2, 1, -1])

lambda_ridge = 1

def squared_exponential_kernel(x, x_prime):
    return np.exp(-0.5 * np.linalg.norm(x - x_prime)**2)

n_samples = X_train.shape[0]
K = np.zeros((n_samples, n_samples))

for i in range(n_samples):
    for j in range(n_samples):
        K[i, j] = squared_exponential_kernel(X_train[i], X_train[j])
```

```python
print(K)
```

```python
I = np.eye(n_samples)
alpha = np.linalg.inv(K + lambda_ridge * I) @ Y_train
```

```python
print(alpha)
```

```python
x_new = np.array([0, 0])
k_new = np.array([squared_exponential_kernel(x_new, X_train[i]) for i in range(n_samples)])

y_new_pred = k_new.T @ alpha
```
